# YOLO

```python
class PerceptionPipeline:
    """
    YOLO-based perception pipeline  to consume CARLA/SUMO ground-truth
    metadata alongside RGB detections.

    The core idea: YOLO alone gives you rough estimates, but when you have simulator data,
    you use it to correct and enrich those estimates before feeding everything into
    a Graph Neural Network (GNN) for risk classification.
    """

    # ------------------------------------------------------------------ init
    def __init__(self,
                 model_path: str = 'yolov8n.pt',
                 conf_threshold: float = 0.4,
                 camera_config: Optional[Dict] = None):

        self.model            = YOLO(model_path)
        self.conf_threshold   = conf_threshold
        self.camera_config    = camera_config or self._default_camera_config()

        self.yolo_to_actor_type = {
            'car':           'vehicle',
            'truck':         'vehicle',
            'bus':           'vehicle',
            'person':        'pedestrian',
            'bicycle':       'cyclist',
            'motorcycle':    'cyclist',
            'traffic light': 'traffic_light',
            'stop sign':     'traffic_sign',
        }
        print(f"[PerceptionPipeline] model={model_path}  conf≥{conf_threshold}")

    # --------------------------------------------------------- camera config
    def _default_camera_config(self) -> Dict:
        """
          This is necessary to project pixel coordinates into real-world metric distances.
          The config matches the simulator's virtual sensor exactly: 800×600 resolution, 90° FOV, mounted at 2.4m height.
          Consistency between simulation and perception is critical — a mismatch here would corrupt all 3D estimates.
        """

        fov_r = math.radians(90)
        w, h  = 800, 600
        fx    = (w / 2.0) / math.tan(fov_r / 2)
        return {
            'focal_length':  fx,
            'image_width':   w,
            'image_height':  h,
            'camera_height': 2.4,
            'pitch_angle':   0.0,
            'x_scale':       40.0,
            'y_scale_min':   5.0,
            'y_scale_max':   50.0,
        }

    # ------------------------------------------------------- 3-D fallback
    def estimate_3d_position(self, bbox: Dict,
                              img_shape: Tuple[int, int]) -> Dict:
        """
        Heuristic 2-D→3-D used ONLY when no metadata GT is available.
        When metadata is present the GT world coords are used directly.
        """
        x_norm      = bbox['x_center']
        y_norm      = bbox['y_center']
        width_norm  = bbox['width']

        x_world = (x_norm - 0.5) * self.camera_config['x_scale']
        y_world = (self.camera_config['y_scale_min'] +
                   (1.0 - y_norm) * self.camera_config['y_scale_max'])
        z_world = 0.0

        aspect_ratio = width_norm / (bbox['height'] + 1e-6)  # kept for future use
        return {
            'x':        float(x_world),
            'y':        float(y_world),
            'z':        float(z_world),
            'yaw':      0.0,
            'speed_ms': 0.0,
            'vx':       0.0,
            'vy':       0.0,
        }

    # ----------------------------------------- match YOLO → GT actor
    @staticmethod
    def _match_gt_actor(yolo_x: float, yolo_y: float,
                        gt_actors: List[Dict],
                        threshold_m: float = 8.0) -> Optional[Dict]:
        """
        For every YOLO detection, it searches all ground-truth actors from the simulator and finds the closest one.
        If within 8 metres → considers them the same object.
        Why 8 metres? Wide enough to tolerate heuristic projection error, narrow enough to avoid wrong matches in dense traffic.
        """
        best, best_d = None, threshold_m
        for actor in gt_actors:
            loc  = actor.get('loc', {})
            dx   = loc.get('x', 0) - yolo_x
            dy   = loc.get('y', 0) - yolo_y
            dist = math.sqrt(dx * dx + dy * dy)
            if dist < best_d:
                best, best_d = actor, dist
        return best

    # --------------------------------------------- load metadata JSON
    @staticmethod
    def load_metadata(meta_path: str) -> Optional[Dict]:
        try:
            with open(meta_path) as f:
                return json.load(f)
        except Exception as e:
            print(f"[WARN] Could not load metadata {meta_path}: {e}")
            return None

    # ================================================== process_frame
    def process_frame(self,
                      frame:         Union[str, np.ndarray],
                      frame_id:      int,
                      run_id:        str  = "perception_run",
                      metadata:      Optional[Dict] = None,
                      target_label:  Optional[int]  = None
                      ) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
        """
        Process one RGB frame.

        Args:
            frame        : image path or numpy array
            frame_id     : integer frame index
            run_id       : dataset run identifier
            metadata     : parsed JSON dict from simulation (preferred)
                           If None, falls back to YOLO-only heuristics.
            target_label : override ground-truth risk label.
                           If None and metadata present, derived from
                           metadata['risk']['level'].

        Returns:
            nodes_df  – one row per detected actor
            edges_df  – pairwise distances + TTC
            parsed_df – frame-level features (matches GNN training schema)
        """

        # ── 0. pull ego ground truth from metadata ─────────────────────────
        if metadata:
            ego_info  = metadata.get('ego', {})
            ego_loc   = ego_info.get('location', {'x': 0, 'y': 0, 'z': 0})
            ego_spd   = ego_info.get('speed_kmh', 0.0) / 3.6        # → m/s
            ego_vel   = ego_info.get('velocity',  {'x': 0.0, 'y': 0.0})
            ego_vx    = ego_vel.get('x', 0.0)
            ego_vy    = ego_vel.get('y', 0.0)
            ego_yaw   = ego_info.get('rotation', {}).get('yaw', 0.0)

            gt_risk   = metadata.get('risk', {})
            gt_vehicles   = metadata.get('vehicles',    [])
            gt_peds       = metadata.get('pedestrians', [])
            gt_all_actors = gt_vehicles + gt_peds
            scenario      = metadata.get('scenario', 'normal')

            # derive target label from risk level if not explicitly given
            if target_label is None:
                level_map = {'safe': 0, 'medium': 1, 'high': 2, 'critical': 3}
                target_label = level_map.get(gt_risk.get('level', 'safe'), 0)
        else:
            ego_spd = ego_vx = ego_vy = 0.0
            ego_yaw = 0.0
            ego_loc = {'x': 0, 'y': 0, 'z': 0}
            gt_risk = {}
            gt_all_actors = []
            scenario = 'normal'
            if target_label is None:
                target_label = 0

        # ── 1. YOLO detection ───────────────────────────────────────────────
        results  = self.model.predict(source=frame,
                                      conf=self.conf_threshold,
                                      verbose=False)[0]
        boxes      = results.boxes
        img_shape  = results.orig_shape   # (H, W)

        # ── 2. Build nodes DataFrame ────────────────────────────────────────
        nodes_data = []
        for i in range(len(boxes)):
            class_id   = int(boxes.cls[i])
            class_name = results.names[class_id]
            confidence = float(boxes.conf[i])
            actor_type = self.yolo_to_actor_type.get(class_name, 'unknown')

            bbox = {
                'x_center': float(boxes.xywhn[i][0]),
                'y_center': float(boxes.xywhn[i][1]),
                'width':    float(boxes.xywhn[i][2]),
                'height':   float(boxes.xywhn[i][3]),
            }

            # Heuristic 3-D estimate (used for GT matching distance calc)
            est_pos  = self.estimate_3d_position(bbox, img_shape)

            # Convert ego-relative heuristic to approximate world coords
            approx_world_x = ego_loc['x'] + est_pos['y'] * math.cos(math.radians(ego_yaw)) \
                           - est_pos['x'] * math.sin(math.radians(ego_yaw))
            approx_world_y = ego_loc['y'] + est_pos['y'] * math.sin(math.radians(ego_yaw)) \
                           + est_pos['x'] * math.cos(math.radians(ego_yaw))

            # Try to fuse with nearest GT actor
            gt_match = self._match_gt_actor(approx_world_x, approx_world_y,
                                             gt_all_actors) if gt_all_actors else None

            if gt_match:
                gt_loc = gt_match.get('loc', {})
                # Use world coords relative to ego for the node features
                rad  = math.radians(ego_yaw)
                dx   = gt_loc.get('x', approx_world_x) - ego_loc['x']
                dy   = gt_loc.get('y', approx_world_y) - ego_loc['y']
                # Rotate to ego-relative frame
                x_ego =  dx * math.cos(-rad) - dy * math.sin(-rad)
                y_ego =  dx * math.sin(-rad) + dy * math.cos(-rad)

                pos = {
                    'x':        round(x_ego, 3),
                    'y':        round(y_ego, 3),
                    'z':        round(gt_loc.get('z', 0.0), 3),
                    'yaw':      round(gt_match.get('yaw', 0.0), 3),
                    'speed_ms': round(gt_match.get('speed_ms', 0.0), 3),
                    'vx':       round(gt_match.get('velocity', {}).get('x', 0.0), 3),
                    'vy':       round(gt_match.get('velocity', {}).get('y', 0.0), 3),
                }
                gt_actor_type = gt_match.get('type', actor_type)  # prefer GT type
                actor_id_src  = gt_match.get('id', f'actor_{frame_id}_{i}')
            else:
                pos           = est_pos
                gt_actor_type = actor_type
                actor_id_src  = f'actor_{frame_id}_{i}'

            node = {
                'actor_id':        f'{actor_id_src}',
                'actor_type':      gt_actor_type,
                'yolo_class':      class_name,
                'frame_id':        frame_id,
                'run_id':          run_id,
                'confidence':      round(confidence, 4),
                'gt_fused':        gt_match is not None,
                **pos,
                # raw YOLO bbox (normalised)
                'bbox_cx':         round(bbox['x_center'], 4),
                'bbox_cy':         round(bbox['y_center'], 4),
                'bbox_w':          round(bbox['width'],    4),
                'bbox_h':          round(bbox['height'],   4),
            }
            nodes_data.append(node)

        # Also add any GT actors NOT detected by YOLO (occluded / OOV)
        if gt_all_actors:
            detected_ids = {n['actor_id'] for n in nodes_data}
            rad = math.radians(ego_yaw)
            for gt_actor in gt_all_actors:
                aid = gt_actor.get('id', '')
                if aid in detected_ids:
                    continue
                gt_loc = gt_actor.get('loc', {})
                dx = gt_loc.get('x', 0) - ego_loc['x']
                dy = gt_loc.get('y', 0) - ego_loc['y']
                x_ego =  dx * math.cos(-rad) - dy * math.sin(-rad)
                y_ego =  dx * math.sin(-rad) + dy * math.cos(-rad)
                nodes_data.append({
                    'actor_id':   aid,
                    'actor_type': gt_actor.get('type', 'vehicle'),
                    'yolo_class': None,
                    'frame_id':   frame_id,
                    'run_id':     run_id,
                    'confidence': 0.0,    # not detected by YOLO
                    'gt_fused':   True,
                    'x':          round(x_ego, 3),
                    'y':          round(y_ego, 3),
                    'z':          round(gt_loc.get('z', 0.0), 3),
                    'yaw':        round(gt_actor.get('yaw', 0.0), 3),
                    'speed_ms':   round(gt_actor.get('speed_ms', 0.0), 3),
                    'vx':         round(gt_actor.get('velocity', {}).get('x', 0.0), 3),
                    'vy':         round(gt_actor.get('velocity', {}).get('y', 0.0), 3),
                    'bbox_cx': None, 'bbox_cy': None,
                    'bbox_w':  None, 'bbox_h':  None,
                })

        nodes_df = pd.DataFrame(nodes_data)

        # ── 3. Edges ────────────────────────────────────────────────────────
        edges_df = self._compute_edges(nodes_df)

        # ── 4. Frame-level features ─────────────────────────────────────────
        parsed_df = self._compute_frame_features(
            nodes_df      = nodes_df,
            edges_df      = edges_df,
            frame_id      = frame_id,
            run_id        = run_id,
            ego_speed_ms  = ego_spd,
            target_label  = target_label,
            gt_risk       = gt_risk,           # ← ground-truth from simulation
            scenario      = scenario,
        )

        return nodes_df, edges_df, parsed_df

        # ====================================================== edges
    def _compute_edges(self, nodes_df: pd.DataFrame) -> pd.DataFrame:
        """
        Why model it as edges?
        The GNN treats the scene as a graph. Actors are nodes, relationships are edges.
        TTC and distance are the edge features that tell the GNN how dangerous the interaction
        between two specific actors is.
        """
        if len(nodes_df) < 2:
            return pd.DataFrame(columns=[
                'source_actor', 'target_actor', 'dist_m', 'ttc_s'])

        edges_data = []
        for i, ri in nodes_df.iterrows():
            for j, rj in nodes_df.iterrows():
                if i >= j:
                    continue
                dist_m = math.sqrt(
                    (ri['x'] - rj['x'])**2 +
                    (ri['y'] - rj['y'])**2 +
                    (ri['z'] - rj['z'])**2
                )
                # TTC using actual velocities when available
                rel_vx  = ri.get('vx', 0) - rj.get('vx', 0)
                rel_vy  = ri.get('vy', 0) - rj.get('vy', 0)
                dx, dy  = ri['x'] - rj['x'], ri['y'] - rj['y']
                closing = -(dx * rel_vx + dy * rel_vy) / max(dist_m, 0.1)
                ttc_s   = dist_m / closing if closing > 0.5 else 999.0

                edges_data.append({
                    'source_actor': ri['actor_id'],
                    'target_actor': rj['actor_id'],
                    'dist_m':       round(dist_m, 3),
                    'ttc_s':        round(ttc_s,  3),
                })
        return pd.DataFrame(edges_data)


    # ================================================ frame-level features
    def _compute_frame_features(self,
                                nodes_df:     pd.DataFrame,
                                edges_df:     pd.DataFrame,
                                frame_id:     int,
                                run_id:       str,
                                ego_speed_ms: float,
                                target_label: int,
                                gt_risk:      Dict,
                                scenario:     str) -> pd.DataFrame:

        # Distances from ego (ego is at origin in the ego-relative frame)
        if len(nodes_df) > 0:
            distances        = np.sqrt(nodes_df['x']**2 +
                                       nodes_df['y']**2 +
                                       nodes_df['z']**2)
            min_dist_m       = float(distances.min())
            closest_actor_dist = min_dist_m
            num_threats      = int((distances < 20.0).sum())
        else:
            min_dist_m = closest_actor_dist = 999.0
            num_threats = 0

        if len(edges_df) > 0:
            ttc_min_s = float(edges_df['ttc_s'].min())
            min_ttc   = ttc_min_s
        else:
            ttc_min_s = min_ttc = 999.0

        # Prefer GT values from simulation when available
        gt_score    = gt_risk.get('score',       None)
        gt_ttc      = gt_risk.get('ttc_min_s',   None)
        gt_min_dist = gt_risk.get('min_dist_m',  None)

        return pd.DataFrame([{
            'frame_id':            frame_id,
            'run_id':              run_id,
            'scenario':            scenario,
            # ego
            'ego_speed_ms':        round(ego_speed_ms, 3),
            # YOLO-computed metrics
            'min_dist_m':          min_dist_m,
            'ttc_min_s':           ttc_min_s,
            'num_threats':         num_threats,
            'min_ttc':             min_ttc,
            'closest_actor_dist':  closest_actor_dist,
            'yolo_detections':     len(nodes_df[nodes_df['confidence'] > 0])
                                   if 'confidence' in nodes_df.columns else 0,
            'gt_fused_count':      int(nodes_df['gt_fused'].sum())
                                   if 'gt_fused' in nodes_df.columns else 0,
            # Ground-truth risk from CARLA simulation (use when available)
            'gt_risk_score':       gt_score,
            'gt_ttc_min_s':        gt_ttc,
            'gt_min_dist_m':       gt_min_dist,
            # label
            'target':              target_label,
        }])

    # ====================================================== dataset runner
    def process_dataset(self,
                        dataset_root: str,
                        output_dir:   str  = "outputs",
                        max_runs:     Optional[int] = None
                        ) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
        """
        Process a full CARLA/SUMO dataset that matches the structure:

            dataset_root/
              data_run_1/
                rgb/          ← images  (NNNNNN.png)
                metadata/     ← JSON    (NNNNNN.json)   ← used for GT
              data_run_2/ ...

        Returns combined (nodes_df, edges_df, parsed_df) for all runs.
        """
        os.makedirs(output_dir, exist_ok=True)

        run_dirs = sorted([
            d for d in os.listdir(dataset_root)
            if d.startswith("data_run_") and
               os.path.isdir(os.path.join(dataset_root, d, "rgb"))
        ])

        if max_runs:
            run_dirs = run_dirs[:max_runs]

        print(f"[PerceptionPipeline] Found {len(run_dirs)} runs under {dataset_root}")

        all_nodes, all_edges, all_parsed = [], [], []

        for run_name in run_dirs:
            rgb_dir  = os.path.join(dataset_root, run_name, "rgb")
            meta_dir = os.path.join(dataset_root, run_name, "metadata")

            image_files = sorted([
                f for f in os.listdir(rgb_dir)
                if f.lower().endswith(('.jpg', '.jpeg', '.png'))
            ])
            print(f"  [{run_name}] {len(image_files)} frames")

            for frame_id, img_file in enumerate(image_files):
                img_path  = os.path.join(rgb_dir,  img_file)
                stem      = os.path.splitext(img_file)[0]
                meta_path = os.path.join(meta_dir, f"{stem}.json")

                # Load ground-truth metadata when present
                metadata = self.load_metadata(meta_path) \
                           if os.path.exists(meta_path) else None

                nodes_df, edges_df, parsed_df = self.process_frame(
                    frame        = img_path,
                    frame_id     = frame_id,
                    run_id       = run_name,
                    metadata     = metadata,
                    target_label = None,    # auto-derived from metadata risk level
                )

                all_nodes.append(nodes_df)
                all_edges.append(edges_df)
                all_parsed.append(parsed_df)

                if frame_id % 50 == 0:
                    print(f"    frame {frame_id:4d}/{len(image_files)} "
                          f"nodes={len(nodes_df)} "
                          f"gt_fused={nodes_df['gt_fused'].sum() if 'gt_fused' in nodes_df.columns else 'n/a'}")

        nodes_all  = pd.concat(all_nodes,  ignore_index=True) if all_nodes  else pd.DataFrame()
        edges_all  = pd.concat(all_edges,  ignore_index=True) if all_edges  else pd.DataFrame()
        parsed_all = pd.concat(all_parsed, ignore_index=True) if all_parsed else pd.DataFrame()

        # Save outputs
        nodes_all .to_csv(os.path.join(output_dir, "nodes.csv"),  index=False)
        edges_all .to_csv(os.path.join(output_dir, "edges.csv"),  index=False)
        parsed_all.to_csv(os.path.join(output_dir, "parsed.csv"), index=False)

        print(f"\n[PerceptionPipeline] Done.")
        print(f"  nodes  : {nodes_all.shape}")
        print(f"  edges  : {edges_all.shape}")
        print(f"  parsed : {parsed_all.shape}")
        print(f"  GT fusion rate: "
              f"{100*nodes_all['gt_fused'].mean():.1f}% of nodes"
              if 'gt_fused' in nodes_all.columns else "")

        return nodes_all, edges_all, parsed_all
```

# Data Cleaning

```python
# 1. Start with a clean copy of the features we want from df
columns_to_remove = [
    "frame_id",
    "ttc_min_s",     # Dropped: Only 1 unique value
    "min_ttc",       # Dropped: Only 1 unique value
    "gt_ttc_min_s",  # Dropped: Matching ground truth metric
    "gt_risk_score"
]

# Work directly on a temporary dataframe to maintain time-series integrity per run_id
working_df = df.drop(columns=columns_to_remove, errors='ignore').copy()

# Ensure rows are sorted sequentially by timeline so historical shifts are accurate
# (Assuming your data has a natural frame sequence order per run)
# working_df = working_df.sort_values(by=['run_id', 'some_frame_index_column'])

# ==========================================
# 2. FEATURE ENGINEERING (Kinematic, Spatial & Historical)
# ==========================================

# Core Kinematic Proxies
working_df["distance_to_speed_ratio"] = working_df["min_dist_m"] / (working_df["ego_speed_ms"] + 1e-5) #Captures how much buffer distance exists relative to speed: A small ratio means the vehicle is moving fast but close to another object → high risk.
working_df["closest_actor_to_speed_ratio"] = working_df["closest_actor_dist"] / (working_df["ego_speed_ms"] + 1e-5) # Highlights immediate collision threats when speed is high and distance is short
working_df["threat_density"] = working_df["num_threats"] / (working_df["yolo_detections"] + 1e-5) # Measures how many threats exist compared to total detections:Differentiates between crowded but safe scenes vs. crowded and dangerous ones.
working_df["ego_kinetic_proxy"] = working_df["ego_speed_ms"] ** 2 #Approximates kinetic energy:Higher squared speed means more severe potential collisions, even if distance looks safe.
working_df["space_compression_index"] = working_df["closest_actor_dist"] / (working_df["ego_kinetic_proxy"] + 1e-5) # Detects situations where high energy is “compressed” into small distances → unsafe.
working_df['log_space_compression'] = np.log1p(working_df['space_compression_index']) #Applies a log transform to normalize skewed values.


# Calculate how much the distance dropped since the last frame (Velocity approximation)
working_df['distance_velocity'] = working_df.groupby('run_id')['min_dist_m'].diff(periods=1) # Approximates relative velocity — shrinking distance means approaching hazard.
working_df['speed_delta'] = working_df.groupby('run_id')['ego_speed_ms'].diff(periods=1) # Captures acceleration/deceleration trends, useful for detecting sudden braking.


# Calculate rolling variance over a 5-frame window to capture unstable driving patterns
working_df['distance_rolling_var'] = working_df.groupby('run_id')['min_dist_m'].transform(
    lambda x: x.rolling(5, min_periods=1).var()
)



# ==========================================

# 3. Separate into X and y & Drop identifiers safely
y = working_df["target"].astype(int)
X = working_df.drop(columns=["run_id", "target"], errors='ignore')

# Handle infinite divisions and drop rows with empty values
X = X.replace([np.inf, -np.inf], np.nan)
nan_mask = X.isna().any(axis=1)
X = X.dropna()
y = y.loc[X.index]

# 4. One-Hot Encode Categorical Driving Scenarios
X = pd.get_dummies(X, columns=["scenario"])

# 5. Stratified Split for Severe Class Imbalance
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(f"Final Feature Set for Modeling ({len(X.columns)} features):\n{list(X.columns)}")
print(f"\nTrain shape: {X_train.shape}, Validation shape: {X_test.shape}")
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import (
    recall_score,
    classification_report,
    log_loss
)

# =========================
# SMOTE
# =========================
smote = SMOTE(
    sampling_strategy='not majority',
    random_state=42,
    k_neighbors=3
)

X_train_sm, y_train_sm = smote.fit_resample(
    X_train,
    y_train
)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(pd.Series(y_train_sm).value_counts())

# =========================
# Weights (balanced only)
# =========================
sample_weights_train = compute_sample_weight(
    class_weight='balanced',
    y=y_train_sm
)

estimators_to_try = [50, 100, 150, 200, 250]

best_score = float('-inf')
best_estimators = None
best_model = None

mlflow.set_experiment("XGBoost_SMOTE_Test")

with mlflow.start_run(run_name="SMOTE_Balanced_Run"):

    print("🚀 Starting SMOTE experiment...\n")

    for n in estimators_to_try:

        params = {
            'objective': 'multi:softprob',
            'num_class': 4,
            'missing': np.nan,
            'eval_metric': 'mlogloss',
            'random_state': 42,

            'max_depth': 4,
            'learning_rate': 0.05,
            'n_estimators': n,
            'early_stopping_rounds': 15,

            'subsample': 0.8,
            'colsample_bytree': 0.8,

            'reg_alpha': 0.5,
            'reg_lambda': 1.5,
            'min_child_weight': 3,
        }

        clf = xgb.XGBClassifier(**params)

        clf.fit(
            X_train_sm,
            y_train_sm,
            sample_weight=sample_weights_train,
            eval_set=[(X_test, y_test)],
            verbose=False
        )

        preds = clf.predict(X_test)
        preds_proba = clf.predict_proba(X_test)

        loss = log_loss(y_test, preds_proba)

        macro_recall = recall_score(
            y_test,
            preds,
            average='macro'
        )

        critical_recall = recall_score(
            y_test,
            preds,
            labels=[3],
            average='macro'
        )

        selection_score = (
            0.5 * macro_recall +
            0.5 * critical_recall
        )

        print(
            f"🌲 Trees: {n:3d} | "
            f"Macro Recall: {macro_recall:.4f} | "
            f"Critical Recall: {critical_recall:.4f}"
        )

        if selection_score > best_score:
            best_score = selection_score
            best_estimators = n
            best_model = clf

    print(f"\n🏁 Best n_estimators = {best_estimators}")

    preds = best_model.predict(X_test)

    print(
        classification_report(
            y_test,
            preds,
            target_names=[
                'Safe',
                'Medium',
                'High',
                'Critical'
            ]
        )
    )

    # =========================
    # Overfitting Check
    # (compares against the SMOTE-resampled training set,
    #  since that's what the model actually trained on)
    # =========================
    train_preds = best_model.predict(X_train_sm)
    train_macro_recall = recall_score(y_train_sm, train_preds, average='macro')
    test_macro_recall = recall_score(y_test, preds, average='macro')

    print(f"\n🔍 Overfitting Check:")
    print(f"   Train macro recall: {train_macro_recall:.4f}")
    print(f"   Test macro recall:  {test_macro_recall:.4f}")
    print(f"   Gap: {train_macro_recall - test_macro_recall:.4f}  "
          f"({'⚠️ possible overfitting' if (train_macro_recall - test_macro_recall) > 0.15 else '✅ looks fine'})")

    model_filename = "best_xgboost_model.json"
    best_model.save_model(model_filename)
    print(f"✅ Weights successfully saved locally as: '{model_filename}'")

    mlflow.log_artifact(model_filename)
    print("☁️ Model weights uploaded and safely backed up to DagsHub!")
```
